# outlier   handling


```python
def outlier_summary(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df[(df[col] < lower) | (df[col] > upper)]

        print(f"{col}")
        print(f"Outliers: {len(outliers)}")
        print(f"Lower: {lower:.2f}")
        print(f"Upper: {upper:.2f}")
        print("-"*40)
```

```python
#outliar summary for selected columns
cols = [
    "total_taps",
    "successful_taps",
    "miss_taps",
    "tap_accuracy",
    "avg_reaction_time_ms",
    "miss_rate"
]

outlier_summary(df_clean, cols)
```

```python
import pandas as pd

# যাদের Hard আছে
hard_df = df_clean[df_clean["stage"] == "hard"].copy()

# Hard করা User-দের ID
hard_users = hard_df["user_id"].unique()

# যাদের শুধু Easy আছে
easy_only_df = df_clean[
    (df_clean["stage"] == "easy") &
    (~df_clean["user_id"].isin(hard_users))
].copy()

# Final Dataset
final_df = pd.concat([hard_df, easy_only_df], ignore_index=True)

print("Final Dataset Shape:", final_df.shape)

print("\nStage Distribution:")
print(final_df["stage"].value_counts())

print("\nLabel Distribution:")
print(final_df["button_label"].value_counts())
```

```python
import numpy as np

# Reaction Delay Normalize
reaction_delay = np.clip(
    (final_df["avg_reaction_time_ms"] - 200) / 1000,
    0,
    1
)

# New IFI
final_df["IFI"] = (
    0.7 * final_df["miss_rate"] +
    0.3 * reaction_delay
)

print(final_df["IFI"].describe())
```

```python
# Screen Area
final_df["screen_area"] = (
    final_df["screen_width"] *
    final_df["screen_height"]
)

# Aspect Ratio
final_df["aspect_ratio"] = (
    final_df["screen_width"] /
    final_df["screen_height"]
)
```
